###Data Frame

```python

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ==========================================
# 1. CREATE THE STUDENT DATASET
# ==========================================

df = pd.DataFrame({
    "Student_ID": range(1, 21),

    "Math": [
        85, 72, 95, 60, 88, 92, 55, 78, 84, 69,
        91, 63, 80, 74, 97, 58, 86, 76, 93, 67
    ],

    "Programming": [
        90, 65, 88, 70, 76, 95, 60, 82, 79, 73,
        87, 68, 85, 78, 96, 62, 91, 72, 89, 75
    ],

    "English": [
        78, 80, 92, 65, 85, 90, 58, 75, 88, 70,
        94, 72, 81, 76, 95, 61, 84, 79, 92, 68
    ],

    "Study_Hours": [
        10, 5, 12, 3, 8, 15, 2, 7, 9, 4,
        11, 6, 8, 5, 14, 3, 10, 6, 13, 4
    ],

    "Attendance": [
        95, 82, 98, 70, 91, 99, 65, 88, 93, 75,
        97, 80, 90, 84, 99, 68, 94, 86, 98, 77
    ],

    "Assignments_Completed": [
        9, 7, 10, 6, 9, 10, 5, 8, 9, 6,
        10, 7, 8, 8, 10, 5, 9, 7, 10, 6
    ]
})

# ==========================================
# 2. CREATE THE TARGET COLUMNS
# ==========================================

df["Average"] = df[
    ["Math", "Programming", "English"]
].mean(axis=1)

df["Result"] = np.where(
    df["Average"] >= 70,
    "Pass",
    "Needs Support"
)

print(df.head())
```

###show first, last 5 students and number of col and rows


```python
print(df.head())
print(df.tail())
print("The rows and columns are :",df.shape)
```

###list the columns names

```python
print(df.columns)
print(df.info())
print(df.dtypes)
```

###Calculate basic statistics

```python
print("The average study hours are :",df['Study_Hours'].mean())
print("The lowest study hours are :",df['Study_Hours'].min())
print("The Highest study hours are",df['Study_Hours'].max())
print("The average attendance are :",df['Attendance'].mean())
print("The lowest attendance are :",df['Attendance'].min())
print("The highest attendance are :",df['Attendance'].max())
```

###Count the result categories

```python
print(df["Result"].value_counts())
print("-"*50)
print(df["Result"].value_counts(normalize=True))
```

###Select the model's input columns

```python
x = df[["Study_Hours", "Attendance", "Assignments_Completed"]]
print(x.head())
```

###Select the target

```python
y = df["Result"]
print(y.head())
print(y.shape)
print(y.unique())
```

###Create training and testing sets

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.25,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

###Understand the split sizes

```python
print("Training rows:", len(x_train))
print("Testing rows:", len(x_test))
```

###Repeat the split with a different test size

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

###Preserve category proportions

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
print(y_train.value_counts(normalize=True))
print("-"*50)
print(y_test.value_counts(normalize=True))
```

###Create the model

```python
classifier = LogisticRegression()
print(classifier)
```

###Train the model

```python
classifier.fit(x_train, y_train)
```

###Make predictions

```python
y_pred = classifier.predict(x_test)
print(y_pred)
```

###Compare actual and predicted results

```python
print("Actual results: " ,y_test.to_list())
print("-"*50)
print("Predicted results: " , y_pred)
```

###Predict for a 1 new student

```python
new_stud = pd.DataFrame({
    "Study_Hours": [8],
    "Attendance": [90],
    "Assignments_Completed": [9]
})
pred_new_stud = classifier.predict(new_stud)
print(pred_new_stud)
```

###Predict for a 3 new students

```python
studs = pd.DataFrame({
    "Study_Hours": [10, 5, 12],
    "Attendance": [95, 82, 98],
    "Assignments_Completed": [9, 7, 10]
})
pred_studs = classifier.predict(studs)
print(pred_studs)
```

###Change a model setting

```python
classifi = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
print(classifi)
classifi.fit(x_train, y_train)
y_preds = classifi.predict(x_test)
print(y_preds)
```
